```python
import numpy as np
from statsmodels.tsa.stattools import adfuller

y = np.array([120, 125, 130, 140, 150, 160, 172, 180, 190, 205])
result = adfuller(y)
print(f"ADF Statistic: {result[0]:.6f}, p-value: {result[1]:.6f}, lags: {result[2]}, nobs: {result[3]}, crit: {result[4]}")

p = 0.05  # уровень значимости 5%
if result[1] < p:
    print("\nВывод: Ряд стационарный")
else:
    print("\nВывод: нестационарный")
```
