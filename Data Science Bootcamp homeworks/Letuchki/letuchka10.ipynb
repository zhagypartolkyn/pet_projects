```python
import numpy as np
import pandas as pd

#pearson
x = [0, 2, 4, 5, 7, 8, 10]
y = [4, 3, 1, 1, 2, 1, 0]
#pearson
x_avg = sum(x)/len(x)
y_avg = sum(y)/len(y)

cov = sum((a - x_avg) * (b - y_avg) for (a,b) in zip(x,y)) / len(x)
stdevX = (sum((a - x_avg)**2 for a in x)/len(x))**0.5
stdevY = (sum((b - y_avg)**2 for b in y)/len(y))**0.5

res = cov/(stdevX*stdevY)
res
```

```python


# Преобразование в DataFrame для упрощения ранжирования
data = pd.DataFrame({'x': x, 'y': y})

data['rank_x'] = data['x'].rank(method='average')
data['rank_y'] = data['y'].rank(method='average')

data['d'] = data['rank_x'] - data['rank_y']
data['d_squared'] = data['d'] ** 2

n = len(x)
spearman_corr = 1 - (6 * data['d_squared'].sum()) / (n * (n**2 - 1))

print("Коэффициент корреляции Спирмена:", spearman_corr)
```
