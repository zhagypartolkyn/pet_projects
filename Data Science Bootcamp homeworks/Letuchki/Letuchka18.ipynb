```python
import torch
import torch.nn as nn
import torch.nn.functional as F

class AttentionMechanism(nn.Module):
    def __init__(self, input_dim, dq, dk, dv):
        super(AttentionMechanism, self).__init__()
        self.Wq = nn.Linear(input_dim, dq, bias=False)
        self.Wk = nn.Linear(input_dim, dk, bias=False)
        self.Wv = nn.Linear(input_dim, dv, bias=False)
    def forward(self, X):
        Q = self.Wq(X)  
        K = self.Wk(X) 
        V = self.Wv(X)
        attention_scores = torch.matmul(Q, K.transpose(-2, -1)) / (K.size(-1) ** 0.5) 
        attention_weights = F.softmax(attention_scores, dim=-1)
        Z = torch.matmul(attention_weights, V)
        return Z, attention_weights
if __name__ == "__main__":
    batch_size = 2
    seq_len = 4 
    input_dim = X.size(-1)
    dq = dk = dv = 16  
    attention = AttentionMechanism(input_dim=input_dim, dq=dq, dk=dk, dv=dv)
    Z, attention_weights = attention(X)
    print("Output (Z):", Z)
    print("Attention Weights:", attention_weights)
```
