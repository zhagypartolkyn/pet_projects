```python
# Замена выбросов медианой


columns_with_log_for_boxplots = ['log_transformed_total_rooms',
        'log_transformed_total_bedrooms', 'log_transformed_population',
        'log_transformed_households']
for col in columns_with_log_for_boxplots:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    median_value = df[col].median()
 
    df[col] = np.where(
        (df[col] < lower_bound) | (df[col] > upper_bound), 
        median_value, 
        df[col]
)
```

### Задание 2. Разделение данных и обучение моделей

**2.1.** Разделите данные на матрицу признаков $X$ и целевую переменную $y$.

```python
# Ваш код:
import statsmodels.api as sm

# удаляю для Х признаки которые показали высокую корреляцию и целевую переменную
X = df_for_modeling.drop(columns = ['log_transformed_total_rooms_sc', 'log_transformed_total_bedrooms_sc',
                                   'log_transformed_households_sc', 'median_house_value_sc'])
y = df_for_modeling['median_house_value_sc']
X = sm.add_constant(X)
```

**2.2.** Разделите данные на тренировочный (80%) и тестовый (20%) наборы.

```python
# Ваш код:
from sklearn.model_selection import train_test_split

train_size = 0.8  # 80% of the data for training

test_size = 0.2  # 20% of the data for testing

train_X, test_X, train_Y, test_Y = train_test_split(X, y, test_size = test_size, 
                                         train_size = train_size, random_state=42)
```

**2.3.** Обучите модель линейной регрессии на тренировочном наборе.

Оцените модель на тестовой выборке, используя метрики:
- Коэффициент детерминации ($R^2$).
- Среднеквадратическая ошибка ($RMSE$).


```python
# Ваш код:
model = sm.regression.linear_model.OLS(train_Y, train_X).fit()
print(model.summary())
```

```python
from sklearn.metrics import mean_squared_error, root_mean_squared_error

print('RMSE_TRAIN', np.sqrt(mean_squared_error(train_Y, model.predict(train_X))))
print('RMSE_TEST: ', np.sqrt(mean_squared_error(test_Y, model.predict(test_X)))) 
```

```python
test_X = sm.add_constant(test_X)
model = sm.regression.linear_model.OLS(test_Y, test_X).fit()
print(model.summary())
```

```python
std_target = df_for_modeling['median_house_value_sc'].std()
print(f"Standard Deviation: {std_target}")
```

**2.5.** Обучите модель Ridge на тренировочной выборке.
- Подберите гиперпараметр alpha через кросс-валидацию.
- Оцените модель с использованием $R^2$ и $RMSE$.
- Объясните, как регуляризация уменьшает величину коэффициентов и предотвращает переобучение.


```python
# Ваш код:

from sklearn.linear_model import Ridge
from sklearn.model_selection import GridSearchCV
from sklearn.linear_model import RidgeCV
```

```python
# Ваш код:

# list of alphas to check: 100 values from 0 to 5 with
r_alphas = np.logspace(0, 10, 10)
# initiate the cross validation over alphas
ridge_model = RidgeCV(alphas=r_alphas, scoring='r2')
# fit the model with the best alpha
ridge_model = ridge_model.fit(train_X, train_Y)

list(zip(train_X.columns, ridge_model.coef_))
```

```python
print('Ridge model scoring with R^2:', ridge_model.score(train_X, train_Y))
print('RMSE_Ridge: ', np.sqrt(mean_squared_error(train_Y, ridge_model.predict(train_X)))) 
```

**2.6.** Обучите модель Lasso на тренировочной выборке.
- Подберите гиперпараметр alpha через кросс-валидацию.
- Оцените модель с использованием $R^2$ и $RMSE$.
- Укажите, какие признаки были обнулены, и объясните, как Lasso помогает упростить модель.

```python
# Зададим диапазон для параметра alpha
from sklearn.linear_model import Lasso
from sklearn.model_selection import GridSearchCV

alpha_values = np.logspace(-4, 0, 50)
lasso = Lasso(max_iter=10000, random_state=42)

grid = GridSearchCV(estimator=lasso, 
                    param_grid={'alpha': alpha_values},
                    scoring='neg_mean_squared_error',  
                    cv=5) 

grid.fit(train_X, train_Y)
best_alpha = grid.best_params_['alpha']
print(f"Лучший alpha: {best_alpha}")
```

```python
lasso_best = Lasso(alpha=best_alpha, max_iter=10000, random_state=42)
lasso_best.fit(train_X, train_Y)
```

```python
#оценка модели на тест выборке
from sklearn.metrics import r2_score

y_pred = lasso_best.predict(test_X)

rmse = np.sqrt(mean_squared_error(test_Y, y_pred))
print('RMSE_Lasso:', (rmse))

r2 = r2_score(test_Y, y_pred)
print('Lasso model scoring with R^2:', (r2))
```

```python
lasso_coefficients = pd.Series(lasso_best.coef_, index=test_X.columns)
zeroed_features_lasso = lasso_coefficients[lasso_coefficients == 0].index.tolist()
print(f"Обнулённые признаки (Lasso): {zeroed_features_lasso}")
```

**3.2.** Сравните метрики между моделями

```python
# Ваш код:
def evaluate_model(y_test, y_pred, model_name):
    r2 = r2_score(y_test, y_pred)
    rmse = root_mean_squared_error(y_test, y_pred)
    print(f"{model_name} - R^2: {r2:.4f}, RMSE: {rmse:.4f}")

# Оценка моделей

evaluate_model(test_Y, y_pred_linear, "Linear")
evaluate_model(test_Y, y_pred_ridge, "Ridge")
evaluate_model(test_Y, y_pred_lasso, "Lasso")
evaluate_model(test_Y, y_pred_elasticnet, "ElasticNet")
```

### Задание 4. Модель
Перед тем как обучить модель, вы можете подготовить данные если они этого требуют.

**4.1.** Постройте модель и выполните её обучение.



```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

train_X_scaled = scaler.fit_transform(train_X)
test_X_scaled = scaler.transform(test_X)
```

```python
from sklearn.linear_model import LogisticRegression

logistic_model = LogisticRegression()
logistic_model.fit(train_X_scaled, train_Y)

y_pred = logistic_model.predict(test_X_scaled)
y_pred
```

**4.2.** Попробуйте поэкспериментировать с различными значениями пороги (например, 0.3, 0.5, 0.7) и оцените, как изменение порога влияет на метрики вашей модели (например, точность, полнота, F1-меру).

```python
# Ваш код:
y_proba = logistic_model.predict_proba(train_X_scaled)[:, 1]  # Вероятности класса 1

y_proba.shape
```

```python
from sklearn.metrics import confusion_matrix, classification_report

thresholds = [0.3, 0.5, 0.7]

for threshold in thresholds:
    y_pred_thresh = (y_proba >= threshold).astype(int)
    print(f'Порог: {threshold}')
    print(confusion_matrix(train_Y, y_pred_thresh))
    print(classification_report(train_Y, y_pred_thresh))
```

**4.3.** Отметьте, какой порог даёт наилучшие метрики для вашей задачи. 

```python
# Ваш код:
from sklearn.metrics import precision_score, recall_score, f1_score

precisions = []
recalls = []
f1_scores = []

for threshold in thresholds:
    y_pred = (y_proba >= threshold).astype(int)
    precisions.append(precision_score(train_Y, y_pred))
    recalls.append(recall_score(train_Y, y_pred))
    f1_scores.append(f1_score(train_Y, y_pred))

plt.figure(figsize=(8,6))
plt.plot(thresholds, precisions, label='Точность')
plt.plot(thresholds, recalls, label='Полнота')
plt.plot(thresholds, f1_scores, label='F1-score')
plt.xlabel('Порог')
plt.ylabel('Значение метрики')
plt.title('Влияние порога на метрики')
plt.legend()
plt.show()

# Порог в 0.5 дает наилучший оптимальный результат; при 0.7 высокий f1, но низкий precision, 
# а при 0.3 выcокий precision, но f1 ниже чем в 0.5 и 0.7 
```
