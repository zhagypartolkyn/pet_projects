```python
import numpy as np 
vec1 = np.array([1, -1, 14, 8, 7, -1, 12])
vec2 = np.array([0, -3, 3, 8, 1, 1, -1])
vec3 = np.array([12, 4, -1, 1, 9, 6, 0])

#d(vec1-vec2, vec3)
#d(vec3*2, vec1/2)
#d(vec1*vec3, vec2^2)
```

```python
#1 Cosine Similarity(A, B) = (A · B) / (||A|| * ||B||)

def find_cosine(A, B):
    dot_product = np.dot(A, B)
    magnitude_A = np.linalg.norm(A)
    magnitude_B = np.linalg.norm(B)
    cosine_similarity = dot_product / (magnitude_A * magnitude_B)
    return cosine_similarity

ans1 = round(find_cosine(vec1 - vec2, vec3), 2)
print(f"Cosine Similarity D(vec1-vec2, vec3): {ans1}")
```

```python
#2 Manhattan distance(A, B) =  Σ|Ai – Bi|

from math import sqrt

def manhattan(A, B):
    return sum(abs(val1-val2) for val1, val2 in zip(A,B))
 
ans2 = manhattan(vec3*2, vec1/2)
print(f"Manhattan distance D(vec3*2, vec1/2): {ans2}")
```

```python
#3 Euclidian distance = D = (x2 − x1)^2+(y2 − y1)^2

def Euclidean_Dist(A, B):
    return np.linalg.norm(A-B)

ans3 = round(Euclidean_Dist(vec1*vec3, vec2**2), 2)
print(f"Euclidian distance D(vec1*vec3, vec2^2): {ans3}")
```
