## 3. Наибольший простой множитель
Простые множители числа $13195$ — это $5, 7, 13$ и $29$.

Найдите наибольший простой множитель числа $600851475143$.


```python
# напишите ваш код здесь

factors = []
def find_max_prime(n):
    p = 2
    while n % p == 0: 
        factors.append(p)
        n /= p
    #checking for odd primes from 3 to sqrt(n), потому что прайм не может быть больше чем sqrt(n)
    for i in range(3, int(n ** 0.5), 2):
        while n % i == 0: 
            factors.append(i) 
            n /= i

    return max(factors)


number1 = 13195
number2 = 600851475143
print(find_max_prime(number1))
print(find_max_prime(number2))
```

## 4. Крупнейшее произведение - палиндром
Палиндромное число читается одинаково в обоих направлениях.

Примеры палиндромных чисел:
$$121, 1331, 12321$$

Наибольший палиндром, полученный умножением двух 2-значных чисел, это $9009 = 91 \times 99$.

Найдите наибольший палиндром, который можно получить в результате умножения двух 3-значных чисел.








```python
# напишите ваш код здесь

def is_palin(number: int) -> bool:
    n = str(number)
    if n == n[::-1]:
        return True
    else:
        return False
        
palindromes = []
def find_max_palin(x: int, y: int) -> int:
    for i in range(x, y):
        for j in range(x, y):
            result = i * j
            if is_palin(result):
                palindromes.append(result)
    return max(palindromes)

num1 = 10
num2 = 100
print(find_max_palin(num1, num2))

num3 = 900
num4 = 1000
print(find_max_palin(num3, num4))
```

## 5. Наименьшее кратное

Число $2520$ является наименьшим числом, которое делится на все числа от $1$ до $10$ без остатка.

Найдите наименьшее положительное число, которое делится на все числа от $1$ до $20$ без остатка.

```python
# напишите ваш код здесь
from functools import reduce

def gcd(x: int, y: int) -> int:
    while y:
        x, y = y, x % y
    return x

def lcm(x: int, y: int) -> int:
    return x * y // gcd(x, y)

def find_lcm_of_all(numbers):
    return reduce(lambda x, y: lcm(x, y), numbers)

numbers= range(1, 20)

print(find_lcm_of_all(numbers))            
```

```python
assert find_lcm_of_all(range(1, 10)) == 2520, "Fail"
```

## 6. Квадратичная разница сумм

Сумма квадратов первых десяти натуральных чисел равна:
$$1^2 + 2^2 + ... + 10^2 = 385.$$

Квадрат суммы первых десяти натуральных чисел равен:
$$(1 + 2 + ... + 10)^2 = 55^2 = 3025.$$

Следовательно, разница между суммой квадратов первых десяти натуральных чисел и квадратом суммы равна:
 $$3025 - 385 = 2640$$

Найдите разницу между суммой квадратов первых ста натуральных чисел и квадратом суммы.


```python
# напишите ваш код здесь
list_of_squares = []

def find_sum_of_squares(n: int) -> int:
    total = 0
    for i in range(1, n+1):
        i = i ** 2
        list_of_squares.append(i)
        total += i
    return total

def find_square_of_sum(n: int) -> int:
    total = 0
    for i in range(1, n+1):
        total += i
    return total ** 2

def find_result(m):
    result = find_square_of_sum(m) - find_sum_of_squares(m)
    return result
    

print(find_result(100))
```

## 7. 10,001-ое простое число

Перечисляя первые шесть простых числа: $2, 3, 5, 7, 11$ и $13$, мы можем увидеть, что $6$-е простое число — это $13$.

Найдите $10,001$-е простое число.

```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

def find_prime_at(limit: int, index: int) -> int:
    prime_numbers = []
    
    for i in range(limit):
        if is_prime(i):
            prime_numbers.append(i)
    
    # Проверка - находится ли индекс в допустимых пределах 
    if 1 <= index <= len(prime_numbers):
        return prime_numbers[index - 1]
    else:
        raise IndexError("Индекс выходит за пределы списка простых чисел.")

m = 20
print(find_prime_at(m, 6))  

m = 200000
print(find_prime_at(m, 10001)) 
```

## 9. Пифагоровы тройки

Пифагоровой тройкой называется набор из трех натуральных чисел $a < b < c $, для которых выполняется равенство:

$$ a^2 + b^2 = c^2. $$

Например, $3^2 + 4^2 = 9 + 16 = 25 = 5^2$

Существует ровно одна Пифагорова тройка, для которой $a + b + c = 1000$  
Найдите произведение $abc$.


```python
# напишите ваш код здесь
import math
summa = 1000
product = 0
def find_abc(sum):
    for a in range(sum):
        for b in range(sum):
            for c in range(sum):
                if (a + b + c == 1000) and (math.pow(a, 2) + math.pow(b, 2) == math.pow(c, 2)) and a < b < c:
                    product = a * b * c
                    return f'a: {a}, b: {b}, c: {c}.\nProduct: {product}'
     
print(find_abc(summa))
```

## 10. Сумма простых чисел
Сумма простых чисел, меньших $10$, равна $2 + 3 + 5 + 7 = 17$.

Найдите сумму всех простых чисел, меньших двух миллионов.

## 12. Треугольное число с высокой степенью делимости

Последовательность треугольных чисел создается путем сложения натуральных чисел. Таким образом, $7$-ое треугольное число будет равно $1 + 2 + 3 + 4 + 5 + 6 + 7 = 28$. Первые десять членов последовательности:

$$1, 3, 6, 10, 15, 21, 28, 36, 45, 55, \dots$$

Давайте перечислим делители первых семи треугольных чисел:

\begin{align}
\mathbf{1} & : 1 \\
\mathbf{3} & : 1, 3 \\
\mathbf{6} & : 1, 2, 3, 6 \\
\mathbf{10} & : 1, 2, 5, 10 \\
\mathbf{15} & : 1, 3, 5, 15 \\
\mathbf{21} & : 1, 3, 7, 21 \\
\mathbf{28} & : 1, 2, 4, 7, 14, 28
\end{align}

Мы можем видеть, что $28$ — это первое треугольное число, имеющее более пяти делителей.

Каково значение первого треугольного числа, имеющего более пятисот делителей?


```python
# напишите ваш код здесь
#formula of triangular number = T = (n*(n+1))/2
def count_divisors(number):
    count = 0
    for i in range(1, int(number ** 0.5) + 1):
        #находим делителей
        if number % i == 0:
            count += 1
            if i != number // i:
                count += 1
    return count

def first_triangle(divisor_limit):
    n = 1
    while True:
        triangle_number = n * (n + 1) // 2
        # Считаем количество делителей
        if count_divisors(triangle_number) > divisor_limit:
            return triangle_number
        n += 1

limit = 500
result = first_triangle(limit)
print("First triangle with more than", limit, "divisors:", result)
```

```python
assert first_triangle(5) == 28, "Fail"
```

## 14. Длина последовательности Коллатца

Дана следующая итеративная последовательность для множества положительных целых чисел:

$$
n \to \frac{n}{2} \quad \text{(если $n$ четное)}
$$
$$
n \to 3n + 1 \quad \text{(если $n$ нечетное)}
$$

Используя приведенные правила и начиная с числа $13$, мы генерируем следующую последовательность:

$$
13 \to 40 \to 20 \to 10 \to 5 \to 16 \to 8 \to 4 \to 2 \to 1
$$

Эта последовательность (начиная с $13$ и заканчивая на $1$) содержит $10$ членов. Хотя это еще не доказано (проблема Коллатца), предполагается, что все начальные числа заканчиваются на $1$.

**Вопрос:** Какое начальное число, меньше одного миллиона, производит самую длинную цепочку?

**Примечание:** Как только цепочка начинается, ее члены могут превышать миллион.



```python
# напишите ваш код здесь

def collatz_length(n, cache):
    original_n = n
    length = 0
    
    while n != 1:
        if n in cache:
            length += cache[n]
            break
        length += 1
        
        if n % 2 == 0:
            n //= 2
        else:
            n = 3 * n + 1
            
    cache[original_n] = length
    return length

max_length = 0
starting_number = 0
cache = {}

for i in range(1, 1000000):
    current_length = collatz_length(i, cache)
    if current_length > max_length:
        max_length = current_length
        starting_number = i

print(f"First number less than 1000000: {starting_number}. Length of the chain: {max_length}")
```

Начиная из верхнего левого угла сетки размером $2 \times 2$ и двигаясь только вправо и вниз, существует ровно $6$ маршрутов к нижнему правому углу.

Каково количество таких маршрутов через сетку $20 \times 20$?


```python
# напишите ваш код здесь

import math
def factorial(num):
    factorial = 1
    for i in range(1, num + 1):
        factorial *= i
    return factorial
        
#C(2n,n)= (2n)!/(sqrt(n!)) 
def find_number_of_routes(n):
    upper = factorial(2 * n)
    down = factorial(n)** 2
    number_of_routes =  upper//down
    
    return number_of_routes

square = 20
print(f"Number of routes for the sqaure {square} * {square} = {find_number_of_routes(square)}")
```

```python
# напишите ваш код здесь


def find_sum_of_pow(num: int, pow: int) -> int:
    number = num ** pow
    summa = 0
    for i in str(number):
        i = int(i)
        summa += i
    return summa

print(find_sum_of_pow(2, 1000))
```

##  17. Количество букв
<p>Если числа от $1$ до $5$ записываются словами: one, two, three, four, five, то общее количество использованных букв составляет $3 + 3 + 5 + 4 + 4 = 19.$</p> <p>Если все числа от $1$ до $1000$ (one thousand) включительно были записаны словами, сколько букв было бы использовано?</p> <br><p class="note"><b>ПРИМЕЧАНИЕ:</b> Не учитывайте пробелы или дефисы. Например, $342$ (three hundred and forty-two) содержит $23$ буквы, а $115$ (one hundred and fifteen) содержит $20$ букв. Использование "and" при записи чисел соответствует британскому использованию.</p>

```python
# напишите ваш код здесь

def number_to_words(n):
    if n == 1000:
        return "one thousand"

    ones = ["", "one", "two", "three", "four", "five", "six", "seven", "eight", "nine"]
    teens = ["ten", "eleven", "twelve", "thirteen", "fourteen", "fifteen", 
             "sixteen", "seventeen", "eighteen", "nineteen"]
    tens = ["", "", "twenty", "thirty", "forty", "fifty", "sixty", "seventy", 
             "eighty", "ninety"]

    words = ""

    if n > 0:
        words += ones[n]

    elif n >= 10:
        words += teens[n - 10]
        n = 0

    elif n >= 20:
        words += tens[n // 10]
        n %= 10
        if n > 0:
            words += "-"
    elif n >= 100:
        words += ones[n // 100] + " hundred"
        n %= 100
        if n > 0:
            words += " and "

    return words

total_letters = 0

for i in range(1, 1001):
    words = number_to_words(i)
    total_letters += len(words.replace(" ", "").replace("-", ""))

total_letters
```

    <li>Год является високосным в двух случаях: либо его номер кратен 4, но при этом не кратен 100, либо кратен 400. Например, високосными являются годы 2000 и 2004, но годы 1900 и 2018 високосными не являются.</li>
</ul>
<p>Сколько воскресений выпало на первое число месяца в течение двадцатого века (с 1 января 1901 года по 31 декабря 2000 года)?</p>

```python
# напишите ваш код здесь

def is_leap_year(year):
    if (year % 400 == 0) and (year % 100 == 0):
        return True
    elif (year % 4 == 0) and (year % 100 != 0):
        return True
    else: return False

def count_sundays_on_first_of_month(start_year, end_year):
    day_of_week = 0  # 1 января 1900 года - понедельник
    sunday_count = 0

    # Месяцы и количество дней в них
    days_in_month = [31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]
    for year in range(start_year, end_year):
        for month in range(12):
            if day_of_week == 6:  
                sunday_count += 1
            
            days_this_month = days_in_month[month]
            if month == 1 and is_leap_year(year):  
                days_this_month += 1

            day_of_week = (day_of_week + days_this_month) % 7

    return sunday_count

result = count_sundays_on_first_of_month(1901, 2000)
print(result)    
```

## 20. Сумма цифр факториала

<p>$n!$ означает $n \times (n - 1) \times \cdots \times 3 \times 2 \times 1$.</p> <p>Например, $10! = 10 \times 9 \times \cdots \times 3 \times 2 \times 1 = 3628800$, и сумма цифр числа $10!$ равна $3 + 6 + 2 + 8 + 8 + 0 + 0 = 27$.</p> <p>Найдите сумму цифр числа $100!$.</p>







```python
# напишите ваш код здесь

def find_sum_of_factorial(num):
    factorial = 1
    sum = 0
    for i in range(1, num + 1):
        factorial *= i
    
    for i in str(factorial):
        sum += int(i)
        
    return sum
    
print(find_sum_of_factorial(10))
print(find_sum_of_factorial(100))
```

```python
# напишите ваш код здесь

def find_indexes(numbers, target):
    for i in numbers: 
        for j in numbers:
            result = i + j
            if result == target:
                return i, j

nums = [1, 2, 5, 6, 7, 8, 10, 65]
res = 6

print(find_indexes(nums, res))
```
