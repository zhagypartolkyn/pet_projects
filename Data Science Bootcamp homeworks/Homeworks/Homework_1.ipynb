## Арифметика: квадратные уравнения


Дано квадратное уравнение:

 $$ax^2 + bx + c = 0$$

где  $a$, $b$ и $c$ — это известные коэффициенты.

Пример уравнения:

$$3x^2 + 7x - 10 = 0$$

Здесь:
- $a = 3$
- $b = 7$
- $c = -10$

Для уравнений этого вида могут быть два, одно или ноль решений. В этой задаче рассматриваются уравнения, у которых два решения. Эти решения вычисляются по следующим формулам:

$$x_1 = \frac{-b + \sqrt{b^2 - 4ac}}{2a}$$
$$x_2 = \frac{-b - \sqrt{b^2 - 4ac}}{2a}$$

### Ваша задача:
1. Вычислите оба корня уравнения $x_1$ и $x_2$.
2. Сохраните оба решения уравнения в переменные `x_1` и `x_2`.
3. Напечатайте значения корней.

### Алгоритм решения:
1. Найдите дискриминант: $D = b^2 - 4ac$.
2. Возьмите квадратный корень из дискриминанта: $\sqrt{D}$.
3. Найдите оба корня по формулам:
   - $x_1 = \frac{-b + \sqrt{D}}{2a}$
   - $x_2 = \frac{-b - \sqrt{D}}{2a}$
   
### Подсказка:
Для вычисления квадратного корня в Python можно использовать возведение в степень 0.5:
```python
sqrt_val = value ** 0.5


````python
# дано: коэффициенты квадратного уравнения
import math
a = 3
b = 7
c = -10

# напишите ваш код здесь

def find_x(x, y, z):
    discriminant = math.pow(b, 2) - (4 * x * z)
    root_d = math.pow(discriminant, 0.5)
    x_1 = round((-y + root_d)/(2 * x), 2)
    x_2 = round((-y - root_d)/(2 * x), 2)
    print("X1:", x_1)
    print("X2:", x_2)

find_x(a, b, c)
    
````
