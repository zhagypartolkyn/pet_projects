```python
#10 Y = coef1*x1 + coef2*x2 + coef3*x3 + coef4*x4 + coef5*x5

Y = [50,
    60,
    55,
    75,
    48,
    65,
    72,
    52,
    80,
    58,
    70,
    45,
    90,
    53,
    62,
    48,
    85,
    55,
    75,
    43]
X = [[12, 175, 70, 0, 1],
    [16, 180, 75, 0, 1],
    [14, 165, 60, 1, 1],
    [18, 185, 80, 0, 1],
    [12, 170, 68, 0, 1],
    [15, 175, 72, 1, 1],
    [17, 178, 76, 0, 1],
    [13, 162, 65, 1, 1],
    [20, 190, 85, 0, 1],
    [14, 168, 70, 1, 1],
    [16, 177, 74, 0, 1],
    [10, 160, 58, 1, 1],
    [22, 192, 88, 0, 1],
    [13, 171, 69, 1, 1],
    [15, 173, 71, 0, 1],
    [11, 166, 67, 1, 1],
    [19, 188, 82, 0, 1],
    [14, 169, 70, 1, 1],
    [17, 176, 75, 0, 1],
    [10, 162, 57, 1, 1]]

Y = np.array(Y)
X = np.array(X)

x1 = X[:, 0]
x2 = X[:, 1]
x3 = X[:, 2]
x4 = X[:, 3]
x5 = X[:, 4]

```

```python
def simple_ols(X, Y):
  X_transpose = X.T
  X_transpose_X = X_transpose @ X
  if np.linalg.det(X_transpose_X) != 0:
    X_t_X_inv = np.linalg.inv(X_transpose_X)
  else:
    print('Error')
  X_t_X_inv_X_t = X_t_X_inv @ X_transpose
  coefs = X_t_X_inv_X_t @ Y
  return coefs

def find_mse(Y_actual, Y_predicted):
    squares = []
    error = Y_actual - Y_predicted
    squares =[i**2 for i in error]
    summa = 0
    for j in squares:
        summa+=j
    mse = summa/len(Y_actual)
    return mse

coef = simple_ols(X, Y)
Y_predicted = x1* coef[0] + x2* coef[1] + x3 * coef[2] + x4 * coef[3] + x5*coef[4]
mse = find_mse(Y, Y_predicted)

print('Coefs and MSE:', coef, round(mse, 5))
```
